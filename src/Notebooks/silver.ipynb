```python
df = spark.table("retail_dev.bronze.sales")
display(df)
```

```python
from pyspark.sql.functions import col, when

df_transformed = (
    df.withColumn("price", col("price").cast("decimal(10,2)"))
      .withColumnRenamed("_ingest_ts", "ingest_ts")
      .withColumn("quantity", when(col("quantity") == -1, 1).otherwise(col("quantity")))
)

display(df_transformed)
```

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS retail_dev.silver")
df_transformed.write.format("delta").mode("overwrite").option("primaryKey", "order_id").saveAsTable("retail_dev.silver.orders_cleaned")
```
