```python
from pyspark.sql.functions import current_timestamp

df = (
    spark.read
         .option("header", "true")
         .option("inferSchema", "true")
         .csv("/Volumes/retail_dev/raw/source/")
         .withColumn("_ingest_ts", current_timestamp())
)

df.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("retail_dev.bronze.sales")


```
